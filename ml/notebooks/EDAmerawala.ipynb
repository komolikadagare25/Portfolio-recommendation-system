# Exploratory Data Analysis (EDA)

## Explainable AI-Based Portfolio Recommendation System

### Objective

The objective of this notebook is to understand the characteristics of historical Indian stock market data before feature engineering and machine learning.

This analysis aims to:

- Understand the dataset structure
- Check data quality
- Identify missing values
- Generate summary statistics
- Study stock behavior
- Discover patterns useful for feature engineering

The insights obtained from this notebook will guide the development of the risk prediction model and the recommendation engine.


```python
import os
from pathlib import Path

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
pd.set_option("display.max_columns", None)
```

```python
BASE_DIR = Path.cwd().parent

INTERIM_DATA_DIR = BASE_DIR / "data" / "interim"

print("Base Directory :", BASE_DIR)
print("Interim Data   :", INTERIM_DATA_DIR)
```

```python
csv_files = sorted(INTERIM_DATA_DIR.glob("*.csv"))

print(f"Total CSV Files : {len(csv_files)}")

csv_files
```

```python
stock_data = {}

for file in csv_files:
    
    stock_name = file.stem.replace("_NS", "")
    
    df = pd.read_csv(file)
    
    stock_data[stock_name] = df

print(stock_data.keys())
```

```python
stock_data["RELIANCE"].head()
```

```python
combined_df = pd.concat(
    stock_data,
    names=["Stock"]
)

combined_df.reset_index(level=0, inplace=True)

combined_df.head()
```

```python
print("Rows :", combined_df.shape[0])
print("Columns :", combined_df.shape[1])
```

```python
combined_df["Date"] = pd.to_datetime(combined_df["Date"])
```

```python
combined_df.info()
```

```python
combined_df.isnull().sum()
```

```python
combined_df.describe().T
```
