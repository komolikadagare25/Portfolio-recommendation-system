```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
```

```python
from pathlib import Path

DATA_PATH = Path("../data/raw/finance_trends.csv")

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully.")
```

```python
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")
```

```python
missing = df.isnull().sum().sort_values(ascending=False)

missing
```

```python
duplicates = df.duplicated().sum()

print(f"Duplicate Rows : {duplicates}")
```

```python
df.describe().T
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

numeric_columns
```

```python
categorical_columns = df.select_dtypes(include="object").columns.tolist()

categorical_columns
```

```python
for column in categorical_columns:
    print("=" * 70)
    print(column)
    print("=" * 70)
    print(df[column].value_counts(dropna=False))
    print()
```

```python
df[numeric_columns].hist(
    figsize=(16, 10),
    bins=25
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    df[numeric_columns].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["age"],
    bins=20,
    kde=True
)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x="gender"
)

plt.title("Gender Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Objective"
)

plt.xticks(rotation=45)

plt.title("Investment Objective")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Duration"
)

plt.xticks(rotation=45)

plt.title("Investment Duration")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Expect"
)

plt.xticks(rotation=45)

plt.title("Expected Return")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Invest_Monitor"
)

plt.xticks(rotation=45)

plt.title("Investment Monitoring Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Source"
)

plt.xticks(rotation=45)

plt.title("Source of Financial Information")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Avenue"
)

plt.xticks(rotation=45)

plt.title("Preferred Investment Avenue")

plt.show()
```
